```python
import random
import math

def calculate_conflicts(state):
    """
    Calculates the number of attacking pairs of queens (cost function).
    """
    conflicts = 0
    n = len(state)
    for i in range(n):
        for j in range(i + 1, n):
            # Same row or same diagonal
            if state[i] == state[j] or abs(state[i] - state[j]) == abs(i - j):
                conflicts += 1
    return conflicts

def get_random_state(n=8):
    """Generates a random board state."""
    return [random.randint(0, n - 1) for _ in range(n)]

def get_best_neighbor(state):
    """
    Generates all neighbors by moving each queen to every possible row in its column,
    and returns the neighbor with the minimum conflicts.
    """
    n = len(state)
    best_neighbors = []
    min_conflicts = float('inf')

    for col in range(n):
        original_row = state[col]
        for row in range(n):
            if row == original_row:
                continue
            # Create a neighbor
            neighbor = list(state)
            neighbor[col] = row
            conflicts = calculate_conflicts(neighbor)

            if conflicts < min_conflicts:
                min_conflicts = conflicts
                best_neighbors = [neighbor]
            elif conflicts == min_conflicts:
                best_neighbors.append(neighbor)

        state[col] = original_row

    return random.choice(best_neighbors) if best_neighbors else state

def simulated_annealing(max_iterations=10000, initial_temp=100.0, cooling_rate=0.99):
    """
    Solves the N-Queens problem using Simulated Annealing.
    """
    current_state = get_random_state()
    current_cost = calculate_conflicts(current_state)

    temp = initial_temp

    for iteration in range(max_iterations):
        if current_cost == 0:
            return current_state, iteration, True

        # Generate a random neighbor instead of always taking the best,
        # to allow escaping local minima.
        n = len(current_state)
        col = random.randint(0, n - 1)
        row = random.randint(0, n - 1)
        while row == current_state[col]:
            row = random.randint(0, n - 1)

        next_state = list(current_state)
        next_state[col] = row
        next_cost = calculate_conflicts(next_state)

        # Calculate cost difference
        delta_e = next_cost - current_cost

        # If the new state is better, or accepted by the Boltzmann probability
        if delta_e < 0 or random.random() < math.exp(-delta_e / temp):
            current_state = next_state
            current_cost = next_cost

        # Cool down the temperature
        temp *= cooling_rate

        # Prevent division by zero or underflow if temp gets too small
        if temp < 1e-5:
            temp = 1e-5

    return current_state, max_iterations, False

def print_board(state):
    """Prints the 8-Queens board configuration."""
    n = len(state)
    for r in range(n):
        row_str = ""
        for c in range(n):
            if state[c] == r:
                row_str += " Q "
            else:
                row_str += " . "
        print(row_str)
    print("\n" + "="*24 + "\n")

if __name__ == "__main__":
    random.seed(42)  # Optional: for reproducible results

    print("Running Simulated Annealing for 8-Queens...")
    solution, iterations, success = simulated_annealing()

    if success:
        print(f"Solution found in {iterations} iterations!\n")
        print(f"State representation (Column -> Row): {solution}")
        print(f"Final Cost (Conflicts): {calculate_conflicts(solution)}\n")
        print("Board Visualization:")
        print_board(solution)
    else:
        print("Failed to find a solution within the iteration limit. Try running again.")
```
