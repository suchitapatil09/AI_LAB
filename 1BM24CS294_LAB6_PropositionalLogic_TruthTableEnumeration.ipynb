```python
import re

def pl_true(sentence, model):
    try:
        return bool(eval(sentence, {}, model))
    except Exception:
        return False

def get_symbols(kb, alpha):
    combined_expression = f"{kb} {alpha}"
    symbols = sorted(list(set(re.findall(r'\b[A-Za-z]\b', combined_expression))))
    return symbols

def tt_check_all_table(kb, alpha, symbols, model, all_symbols):
    if not symbols:
        kb_eval = pl_true(kb, model)
        alpha_eval = pl_true(alpha, model)

        row = []
        for sym in all_symbols:
            row.append(str(model[sym]).ljust(5))

        row.append(str(kb_eval).ljust(5))
        row.append(str(alpha_eval).ljust(5))

        print(" | ".join(row))

        if kb_eval:
            return alpha_eval
        return True
    else:
        p = symbols[0]
        rest = symbols[1:]

        model_true = model.copy()
        model_true[p] = True

        model_false = model.copy()
        model_false[p] = False

        result_false = tt_check_all_table(kb, alpha, rest, model_false, all_symbols)
        result_true = tt_check_all_table(kb, alpha, rest, model_true, all_symbols)

        return result_true and result_false

def generate_truth_table(kb, alpha, query_name):
    print(f"\n--- Testing Query: {query_name} ---")
    symbols = get_symbols(kb, alpha)

    header = [sym.ljust(5) for sym in symbols] + ["KB   ", "Alpha"]
    header_str = " | ".join(header)

    print(header_str)
    print("-" * len(header_str))

    entails = tt_check_all_table(kb, alpha, symbols, {}, symbols)

    print("-" * len(header_str))
    print(f"Entailment (KB |= Alpha): {entails}")


if __name__ == "__main__":
    # Translating logic to Python syntax:
    # Q -> P becomes (not Q or P)
    # P -> ~Q becomes (not P or not Q)
    # Q V R becomes (Q or R)
    kb = "(not Q or P) and (not P or not Q) and (Q or R)"

    alpha1 = "R"
    generate_truth_table(kb, alpha1, "R")

    alpha2 = "(not R or P)"
    generate_truth_table(kb, alpha2, "R -> P")

    alpha3 = "(not Q or R)"
    generate_truth_table(kb, alpha3, "Q -> R")
```
