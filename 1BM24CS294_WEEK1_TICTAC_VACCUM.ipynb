```python
#TIC TAC TOE
import random

board = [str(i) for i in range(1, 10)]
human = input("Choose X or O: ").upper()
comp = "O" if human == "X" else "X"

for turn in range(9):
    for i in range(0, 9, 3):
        print(" | ".join(board[i : i + 3]))


    player = human if turn % 2 == 0 else comp
    if player == human:
        move = input(f"Your turn ({human}). Pick a cell: ")
        while move not in board:
            move = input("Invalid! Pick an available number: ")
    else:
        move = random.choice([cell for cell in board if cell in "123456789"])
        print(f"Computer ({comp}) picked {move}")

    board[int(move) - 1] = player


    lines = [
        board[0:3],
        board[3:6],
        board[6:9],
        board[0::3],
        board[1::3],
        board[2::3],
        board[0::4],
        board[2::2],
    ]
    if [player] * 3 in lines:
        print(
            "\n" + "\n".join(" | ".join(board[i : i + 3]) for i in range(0, 9, 3))
        )
        print(f"\n{player} wins!")
        break
else:
    print("\nIt's a draw!")
```

```python
#VACCUM CLEANER SIMPLE REFLEX AGENT
import random

def simple_reflex_vacuum_agent(location, status):
    if status == "Dirty":
        return "Suck"
    elif location == "A":
        return "Right"
    elif location == "B":
        return "Left"

environment = {
    "A": random.choice(["Clean", "Dirty"]),
    "B": random.choice(["Clean", "Dirty"])
}
agent_location = random.choice(["A", "B"])

print("--- SIMPLE REFLEX AGENT SIMULATION ---")
print(f"Initial State: {environment}, Agent Location: {agent_location}\n")

for step in range(1, 5):
    status = environment[agent_location]
    action = simple_reflex_vacuum_agent(agent_location, status)
    print(f"Step {step}: Percept=({agent_location}, {status}) -> Action={action}")

    if action == "Suck":
        environment[agent_location] = "Clean"
    elif action == "Right":
        agent_location = "B"
    elif action == "Left":
        agent_location = "A"

print(f"\nFinal State: {environment}")
```
